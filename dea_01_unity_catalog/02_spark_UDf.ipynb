# Custom SQL Functions for Gizmobox

This notebook creates two simple SQL functions that we can reuse anywhere in the workspace. Instead of writing the same CASE statements or string formatting over and over, we define them once here.

## What it creates

**`format_name(firstname, lastname)`**
* Takes two names and returns them properly capitalized and combined
* Example: `format_name('JOHN', 'doe')` gives you `'John Doe'`

**`get_payment_status(payment_status)`**
* Converts status numbers to readable text
* 1 = Success, 2 = Pending, 3 = Cancelled, 4 = Failed
* So `get_payment_status(1)` returns `'Success'`

The last cell shows how to use these functions in a real query on the payments table.

## How to use it

1. Make sure the gizmobox catalog exists (run notebook 01 first)
2. Run all cells here
3. Now you can use these functions in any query: `gizmobox.default.format_name()` or `gizmobox.default.get_payment_status()`

These functions get saved in Unity Catalog so everyone on the team can use them.

```sql
create or replace function gizmobox.default.format_name (firstname string, lastname string) 
returns string  
return CONCAT(initcap(firstname), ' ', initcap(lastname))
```

```sql
create or replace function gizmobox.default.get_payment_status  (payment_status Int) 
returns string 
return case payment_status 
 when 1 then "Success"
 when 2 then "Pending"
 when 3 then "Cancelled"
 when 4 then "Failed" 
end 
```
