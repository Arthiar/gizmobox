### Configure Access to Cloud Storage Data via Unity Catalog

#### Connecting to Access Connector 

```sql
%fs ls 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/landing/operational_data/'
```

```sql
%sql
CREATE EXTERNAL LOCATION if not exists dea_poject_ext_dl_demo 
    URL 'abfss://demo@deaproject1extdl1.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL dea_project_ext_sc)
    COMMENT "Ext location for a demo "
```

```sql
%sql
CREATE EXTERNAL LOCATION if not exists dea_poject_ext_dl_gizmobox 
    URL 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL dea_project_ext_sc)
    COMMENT "Ext location for a demo "
```

## Create catalog gizmobox

```sql
%sql
CREATE CATALOG  IF NOT EXISTS gizmobox
      MANAGED LOCATION 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/'
```

```sql
%sql
use catalog gizmobox
```

```sql
%sql
CREATE SCHEMA  IF NOT EXISTS  landing
      MANAGED LOCATION 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/';
CREATE SCHEMA  IF NOT EXISTS  bronze
      MANAGED LOCATION 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/';
CREATE SCHEMA  IF NOT EXISTS  silver
      MANAGED LOCATION 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/';
CREATE SCHEMA  IF NOT EXISTS  gold
      MANAGED LOCATION 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/';
```

```sql
use catalog gizmobox
```

```sql
SHOW SCHEMAS
```

Create Volume

```sql
USE CATALOG gizmobox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
LOCATION 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/landing/operational_data/'
COMMENT 'External volume for operational data';
```
