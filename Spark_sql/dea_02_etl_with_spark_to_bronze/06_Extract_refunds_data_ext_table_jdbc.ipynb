This notebook creates and fills the bronze refunds table, then shows the loaded refund data.

#### Refunds bronze table walkthrough

This notebook creates a small refunds dataset directly in Databricks, stores it in the bronze layer, and then verifies the loaded records with simple preview queries.

Select the `gizmobox` catalog and create the bronze schema if it does not already exist. This gives the refund table a managed location in Unity Catalog.

```sql
USE CATALOG gizmobox;
CREATE SCHEMA IF NOT EXISTS bronze;
```

```sql
%python
# 1. Connect and pull raw rows from your Azure SQL database
sql_df = spark.read \
    .format("jdbc") \
    .option("url", "jdbc:sqlserver://gizmobox-db1.database.windows.net:1433;database=gizmoboxdb;") \
    .option("dbtable", "dbo.Refunds") \
    .option("user", "gizmoboxadm") \
    .option("password", "Arthi@123456") \
    .load()

# 2. Write the DataFrame directly into your Unity Catalog 3-tier namespace path
sql_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("gizmobox.bronze.refunds")

print("Table successfully written to Unity Catalog!")
```
