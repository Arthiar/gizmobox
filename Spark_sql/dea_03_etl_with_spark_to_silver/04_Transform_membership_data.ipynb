This notebook gets customer IDs from membership file names and creates the silver membership table.

```sql
select * from gizmobox.bronze.v_membership
```

```sql
select regexp_extract(path, '.*/([0-9]+)\\.png$', 1) As customer_id,
content 
from gizmobox.bronze.v_membership
```

```sql
CREATE OR REPLACE TABLE gizmobox.silver.membership
AS 
select regexp_extract(path, '.*/([0-9]+)\\.png$', 1) As customer_id,
content 
from gizmobox.bronze.v_membership
```
