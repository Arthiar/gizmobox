This notebook puts shipping and billing addresses into one row per customer and creates the silver address table.

```sql
select * from gizmobox.bronze.v_addresses
```

```sql
select * from (
    select customer_id, 
    address_type, 
    address_line_1, 
    city,
    state,
    postcode 
    from gizmobox.bronze.v_addresses
)
Pivot 
(Max (address_line_1)as address_line_1,
Max (city)as city,
Max (state)as state,
Max (postcode)as postcode
for address_type in ('shipping','billing'));
```

```sql
create or replace table gizmobox.silver.address
as
select * from (
    select customer_id, 
    address_type, 
    address_line_1, 
    city,
    state,
    postcode 
    from gizmobox.bronze.v_addresses
)
Pivot 
(Max (address_line_1)as address_line_1,
Max (city)as city,
Max (state)as state,
Max (postcode)as postcode
for address_type in ('shipping','billing'));
```

```sql
select * from gizmobox.silver.address
```
