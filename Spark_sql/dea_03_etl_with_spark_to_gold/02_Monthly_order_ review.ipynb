```sql
select date_format(any_value(transaction_timestamp), "yyyy-MM") AS transaction_month,
                   customer_id,
                   count(Distinct(order_id)) as total_orders,
                   sum(quantity) as total_quantity_bought,
                   sum(quantity * price) as total_amount
from gizmobox.silver.orders
group by order_id, customer_id
```

```sql
create or replace table gizmobox.gold.order_monthly_summary 
as 
select date_format(any_value(transaction_timestamp), "yyyy-MM") AS transaction_month,
                   customer_id,
                   count(Distinct(order_id)) as total_orders,
                   sum(quantity) as total_quantity_bought,
                   sum(quantity * price) as total_amount
from gizmobox.silver.orders
group by order_id, customer_id
```
