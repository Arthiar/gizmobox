This notebook puts shipping and billing addresses into one row per customer and creates the silver address table.

```python
df = spark.table("gizmobox.bronze.py_addresses")
display(df)
```

```python
from pyspark.sql.functions import *
```

```python
df_final = (
    df
    .groupBy('customer_id')
    .pivot('address_type', ['billing', 'shipping'])
    .agg(
        max('address_line_1').alias('address_line_1'),
        max('city').alias('city'),
        max('state').alias('state'),
        max('postcode').alias('postcode')
    )
   )
display(df_final)
```

```python
df_final.write.mode("overwrite").saveAsTable("gizmobox.silver.py_addresses")
```
