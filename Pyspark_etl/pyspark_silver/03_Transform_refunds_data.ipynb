This notebook cleans refund data, splits the refund reason and source, and creates the silver refunds table.

```python

df = spark.read.table('gizmobox.bronze.py_refunds')
display(df)
```

```python
from pyspark.sql.functions import *
```

```python
df_specific = (
    df.select(
        "refund_id",
        "payment_id",
        date_format("refund_timestamp", "yyyy-MM-dd").cast('date').alias("refund_date"),
        date_format("refund_timestamp", "HH-mm-ss").alias("refund_time"),
        "refund_amount",
        regexp_extract("refund_reason",'^([^:]+)',1).alias ("refund_reason"),
        regexp_extract("refund_reason",'([^:]+)$',1).alias("refund_reason_description")
    )
)
display(df_specific)
```

```python
df_specific.write.mode("overwrite").saveAsTable("gizmobox.silver.py_refunds")
```
