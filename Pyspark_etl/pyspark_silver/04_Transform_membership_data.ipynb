This notebook gets customer IDs from membership file names and creates the silver membership table.

```python
df = spark.read.table("gizmobox.bronze.py_memberships")
display(df)
```

```python
from pyspark.sql.functions import regexp_extract
```

```python
df_extract = (
    df.select(
        regexp_extract('path', '.*/([0-9]+)\\.png$', 1).alias('customer_id'),
        'content'   )
)
display(df_extract)
```

```python
df_extract.write.mode('overwrite').saveAsTable('gizmobox.silver.py_memberships')
```
