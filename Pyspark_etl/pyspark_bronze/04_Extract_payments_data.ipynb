```python
payments_schema = 'payments_id Integer, order_id Integer, payment_timestamp Timestamp, payment_status Integer, payment_method String'
```

```python
df = (spark.read.format("csv")
      .option("header", "true")
      .option("delimiter", ",")
      .schema(payments_schema)
     .load("abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/landing/external_data/payments"))
display(df)
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("gizmobox.bronze.py_payments")
```
