Add the file path so it can later be used in audit columns 

```sql
select _metadata.file_path AS filepath, * 
from json.`/Volumes/gizmobox/landing/operational_data/customers`
```

#### Register Files in the Unity Catalog using Views

```sql
create or replace view gizmobox.bronze.v_customer 
as
select * ,
_metadata.file_path AS filepath 
from json.`/Volumes/gizmobox/landing/operational_data/customers`
```

```sql
select * from gizmobox.bronze.v_customer
```

```sql
create or replace temporary view tv_customer 
as
select * ,
_metadata.file_path AS filepath 
from json.`/Volumes/gizmobox/landing/operational_data/customers`
```
