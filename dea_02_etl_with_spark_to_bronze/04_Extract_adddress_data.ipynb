Try reading the address files with the default CSV reader first. This quick test shows why the tab delimiter still needs to be supplied explicitly for these TSV files.

```sql
select * from csv.`/Volumes/gizmobox/landing/operational_data/addresses/`
```

Read the address files again with `read_files`, this time using the tab delimiter and header row. This step parses the TSV data into proper columns instead of one raw text field.

```sql
select * from read_files('/Volumes/gizmobox/landing/operational_data/addresses/',
                        format=>'csv',
                        delimiter =>'\t',
                        header=>True)
```

Create a reusable bronze view over the parsed address files so downstream notebooks can query the address data without repeating the file-reading options.

```sql
create or replace view gizmobox.bronze.v_addresses
as 
select * from read_files('/Volumes/gizmobox/landing/operational_data/addresses/',
                        format=>'csv',
                        delimiter =>'\t',
                        header=>True)
```

Query the bronze address view to confirm that the view was created successfully and that the TSV files are exposed as structured columns.

```sql
select * from gizmobox.bronze.v_addresses
```
