List the payment files in external storage to confirm the location and the monthly CSV extracts before creating the external table.

```sql
%fs ls 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/landing/external_data/payments'
```

Create an external bronze table over the payment CSV files so Databricks can query the data directly from the external storage location.

```sql
create table if not exists gizmobox.bronze.payments (
    paymant_id int,
    customer_id int,
    payment_timestamp timestamp,
    payment_status integer,
    payment_method string
)
Using CSV 
Options (
  header = 'true',
  delimiter = ',')
  location 'abfss://gizmobox@deaproject1extdl1.dfs.core.windows.net/landing/external_data/payments'
```

Query the payments table to confirm that the external table definition reads the CSV files correctly and exposes the expected columns.

```sql
select * from gizmobox.bronze.payments
```
