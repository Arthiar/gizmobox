```sql
select distinct(*) 
from 
gizmobox.bronze.v_customer
where customer_id is not null
order by customer_id
```

```sql
create or replace temporary view v_distinct_customers 
as
select distinct(*) 
from 
gizmobox.bronze.v_customer
where customer_id is not null
order by customer_id
```

```sql
select customer_id,
        max(created_timestamp) As max_timestamp
    from gizmobox.bronze.v_customer
    group by customer_id
```

```sql
with cte_max as (
select customer_id,
        max(created_timestamp) As max_timestamp
    from gizmobox.bronze.v_customer
    group by customer_id)
select * 
from v_distinct_customers d
Join cte_max m
on d.customer_id = m.customer_id
and d.created_timestamp = m.max_timestamp
```

```sql
with cte_max as (
select customer_id,
        max(created_timestamp) As max_timestamp
    from gizmobox.bronze.v_customer
    group by customer_id)
select cast(d.created_timestamp as timestamp) AS created_timestamp,
        d.customer_id,
        d.customer_name,
        cast(d.date_of_birth as date) as date_of_birth,
        d.email,
       cast(d.member_since as date) as member_since,
        d.telephone
from v_distinct_customers d
Join cte_max m
on d.customer_id = m.customer_id
and d.created_timestamp = m.max_timestamp
```

```sql
create table if not exists gizmobox.silver.customer
as 
with cte_max as (
select customer_id,
        max(created_timestamp) As max_timestamp
    from gizmobox.bronze.v_customer
    group by customer_id)
select cast(d.created_timestamp as timestamp) AS created_timestamp,
        d.customer_id,
        d.customer_name,
        cast(d.date_of_birth as date) as date_of_birth,
        d.email,
       cast(d.member_since as date) as member_since,
        d.telephone
from v_distinct_customers d
Join cte_max m
on d.customer_id = m.customer_id
and d.created_timestamp = m.max_timestamp
```
