This notebook explodes the order items array into one row per item and creates the silver orders table.

```sql
select * from gizmobox.silver.orders_json
```

```sql
select json_value.customer_id,
json_value.items,
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp 
from gizmobox.silver.orders_json
```

```sql
select json_value.customer_id,
array_distinct(json_value.items) as items,
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp 
from gizmobox.silver.orders_json
```

```sql
select json_value.customer_id,
explode(array_distinct(json_value.items) )as items,
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp 
from gizmobox.silver.orders_json
```

```sql
create or replace temp view tv_orders
as 
select json_value.customer_id,
explode(array_distinct(json_value.items) )as items,
json_value.order_id,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp 
from gizmobox.silver.orders_json
```

```sql
select 
order_id,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
items.item_id,
items.name,
items.price,
items.quantity,
items.category,
items.details.brand,
items.details.color
 from tv_orders
```

```sql
create or replace table gizmobox.silver.orders
as 
select 
order_id,
order_status,
payment_method,
total_amount,
transaction_timestamp,
customer_id,
items.item_id,
items.name,
items.price,
items.quantity,
items.category,
items.details.brand,
items.details.color
 from tv_orders
```
