This notebook cleans payment fields, changes status codes to words, and creates the silver payments table.

```sql
select * from gizmobox.bronze.payments
```

```sql
select paymant_id, 
customer_id, 
cast(date_format(payment_timestamp,"yyyy-MM-dd") as date) as payment_date,
cast (date_format(payment_timestamp,"HH:mm:ss") as timestamp) as payment_time,
payment_status, 
payment_method 
from gizmobox.bronze.payments
```

```sql
select paymant_id, 
customer_id, 
cast(date_format(payment_timestamp,"yyyy-MM-dd") as date) as payment_date,
cast (date_format(payment_timestamp,"HH:mm:ss") as timestamp) as payment_time,
case payment_status 
 when 1 then "Success"
 when 2 then "Pending"
 when 3 then "Cancelled"
 when 4 then "Failed" 
end as payment_status,
payment_method 
from gizmobox.bronze.payments
```

```sql
create table if not exists gizmobox.silver.payments
as
select paymant_id, 
customer_id, 
cast(date_format(payment_timestamp,"yyyy-MM-dd") as date) as payment_date,
cast (date_format(payment_timestamp,"HH:mm:ss") as timestamp) as payment_time,
case payment_status 
 when 1 then "Success"
 when 2 then "Pending"
 when 3 then "Cancelled"
 when 4 then "Failed" 
end as payment_status,
payment_method 
from gizmobox.bronze.payments
```
