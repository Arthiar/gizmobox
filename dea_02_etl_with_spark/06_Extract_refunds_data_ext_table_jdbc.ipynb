#### Refunds bronze table walkthrough

This notebook creates a small refunds dataset directly in Databricks, stores it in the bronze layer, and then verifies the loaded records with simple preview queries.

Select the `gizmobox` catalog and create the bronze schema if it does not already exist. This gives the refund table a managed location in Unity Catalog.

```sql
USE CATALOG gizmobox;
CREATE SCHEMA IF NOT EXISTS bronze;
```

Create the bronze `refunds` table in Delta format and load the sample refund rows into it. This step materializes the dataset inside Databricks so it can be queried like any other managed table.

```sql
CREATE OR REPLACE TABLE gizmobox.bronze.refunds (
  refund_id INT,
  payment_id INT,
  refund_timestamp TIMESTAMP,
  refund_amount DECIMAL(10, 2),
  refund_reason STRING
)
USING DELTA;

INSERT OVERWRITE gizmobox.bronze.refunds
SELECT *
FROM VALUES
  (1, 66, TIMESTAMP('2025-01-10 11:30:00'), CAST(85.75 AS DECIMAL(10,2)), 'Payment Error:Retailer'),
  (2, 69, TIMESTAMP('2025-01-03 12:40:15'), CAST(120.50 AS DECIMAL(10,2)), 'Order Cancelled:Customer'),
  (3, 72, TIMESTAMP('2025-01-06 14:45:30'), CAST(65.00 AS DECIMAL(10,2)), 'Product Returned:Customer'),
  (4, 73, TIMESTAMP('2025-01-07 16:10:45'), CAST(210.99 AS DECIMAL(10,2)), 'Order Cancelled:Customer'),
  (5, 75, TIMESTAMP('2025-01-09 18:25:00'), CAST(45.20 AS DECIMAL(10,2)), 'Payment Error:Retailer'),
  (6, 80, TIMESTAMP('2025-01-10 09:35:20'), CAST(130.15 AS DECIMAL(10,2)), 'Order Cancelled:Customer'),
  (7, 83, TIMESTAMP('2025-01-12 11:20:40'), CAST(150.00 AS DECIMAL(10,2)), 'Product Returned:Customer'),
  (8, 85, TIMESTAMP('2025-01-14 13:15:30'), CAST(89.99 AS DECIMAL(10,2)), 'Order Cancelled:Customer'),
  (9, 89, TIMESTAMP('2025-01-15 15:00:00'), CAST(78.50 AS DECIMAL(10,2)), 'Payment Error:Retailer'),
  (10, 91, TIMESTAMP('2025-01-17 16:45:15'), CAST(250.75 AS DECIMAL(10,2)), 'Product Returned:Customer')
AS refunds(refund_id, payment_id, refund_timestamp, refund_amount, refund_reason);
```

Preview the refunds table in descending timestamp order so the newest refund records appear first. This is a quick validation that the load completed successfully and the key columns look correct.

```sql
SELECT
  refund_id,
  payment_id,
  refund_timestamp,
  refund_amount,
  refund_reason
FROM gizmobox.bronze.refunds
ORDER BY refund_timestamp DESC
LIMIT 20
```
